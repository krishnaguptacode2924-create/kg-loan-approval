```python
df.isnull().sum()
```

```python
Column = ['Loan_ID', 'Gender', 'Married', 'Dependents', 'Education',
          'Self_Employed', 'ApplicantIncome', 'CoapplicantIncome', 'LoanAmount',
          'Loan_Amount_Term', 'Credit_History', 'Property_Area', 'Loan_Status']
```

8. **.value_counts()** counts how many times each unique value appears in a Series (usually a DataFrame column). It's useful for understanding the distribution of values in your dataset.

```python
for i in Column[1:]:
    print(df[i].value_counts(dropna=False))
```

```python
# 1. Continuous Numerical Features (Distribution Plot with KDE)
num_cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount']

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

for ax, col in zip(axes, num_cols):
    sns.histplot(df[col], kde=True, bins=30, ax=ax, color='skyblue')
    ax.set_title(f'Distribution of {col}')

plt.tight_layout()
plt.show()


# 2. Categorical & Discrete Features (Count Plots - No KDE)

cat_cols = [
    'Gender', 'Married', 'Dependents', 'Education',
    'Self_Employed', 'Loan_Amount_Term', 'Credit_History',
    'Property_Area', 'Loan_Status'
]

fig, axes = plt.subplots(3, 3, figsize=(15, 12))

for ax, col in zip(axes.flatten(), cat_cols):
    sns.countplot(data=df, x=col, ax=ax, palette='Set2')
    ax.set_title(f'Count of {col}')

plt.tight_layout()
plt.show()
```
