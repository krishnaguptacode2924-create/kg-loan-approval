# Data Preprocessing

**Import Modules**

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

**Import Dataset**

```python
data_url = "https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/Loan-Approval-Prediction.csv"
```

```python
df = pd.read_csv(data_url)
df
```

Keep clean data in a separate copy to preserve original data

```python
df_clean = df.copy()
df_clean
```

**.drop_duplicates** removes duplicate values and inplace=True makes it permanent

```python
df_clean.drop_duplicates(inplace = True)
df_clean.shape
```

**For Gender column:**

1. Fill null values with mode
2. convert Male/Female = 0/1
3. Change column name to Female

```python
df_clean['Gender'] = df_clean['Gender'].fillna(df_clean['Gender'].mode()[0])
df_clean ['Gender'] = df_clean['Gender'].map({"Male" : 0,"Female" : 1})
df_clean.rename (columns={'Gender' : 'Female'}, inplace = True)
df_clean
```

**For Married column:**

1. Fill null values with mode
2. convert No/Yes = 0/1

```python
df_clean['Married'] = df_clean['Married'].fillna(df_clean['Married'].mode()[0])
df_clean ['Married'] = df_clean['Married'].map({"No" : 0,"Yes" : 1})
df_clean
```

**For Education column:**

1. convert Not Graduate/Graduate = 0/1
2. Change column name to Education

```python
df_clean ['Education'] = df_clean['Education'].map({"Not Graduate" : 0,"Graduate" : 1})
df_clean.rename (columns={'Education' : 'Graduate'}, inplace = True)
df_clean
```

**For Self_Employed column:**

1. Fill null values with mode
2. convert No/Yes = 0/1

```python
df_clean['Self_Employed'] = df_clean['Self_Employed'].fillna(df_clean['Self_Employed'].mode()[0])
df_clean['Self_Employed'] = df_clean['Self_Employed'].map({"No" : 0,"Yes" : 1})
df_clean
```

**For Loan_Status column:**

1. convert N/Y = 0/1
2. Change column name to Approved

```python
df_clean['Loan_Status'] = df_clean['Loan_Status'].map({"N" : 0,"Y" : 1})
df_clean.rename (columns={'Loan_Status' : 'Approved'}, inplace = True)
df_clean
```

**For Property_Area column:**

Convert it into different columns

```python
df_clean = pd.get_dummies(df_clean, columns = ['Property_Area'],drop_first=False)
df_clean.head()
```

For each new column map True/False = 1/0

```python
colem = ['Property_Area_Rural', 'Property_Area_Semiurban', 'Property_Area_Urban']

for c in colem:
    df_clean[c] = df_clean[c].astype(int)

df_clean.head()
```

```python
for j in colem:
    df_clean.rename (columns={j : j[14:]}, inplace = True)

df_clean.head()
```

Drop **Loan_ID** column as it has no contribution to approval/rejection

```python
df_clean.drop(columns=['Loan_ID'], inplace=True)
df_clean.head(0)
```

```python
df_clean.isnull().sum()
```

**For Dependents column:**

1. Fill null values with mode
2. convert string values ti integer

```python
df_clean['Dependents'] = df_clean['Dependents'].fillna(df_clean['Dependents'].mode()[0])
df_clean['Dependents'] = df_clean['Dependents'].map({"1" : 1,"2" : 2, "3+": 3, "0": 0})
df_clean.head()
```

**For Loan_Amount_Term column:**

1. Fill null values with mode
2. Change column name to Term

```python
df_clean['Loan_Amount_Term'] = df_clean['Loan_Amount_Term'].fillna(df_clean['Loan_Amount_Term'].mode()[0]).astype(int)
df_clean.rename (columns={'Loan_Amount_Term' : 'Term'}, inplace = True)
df_clean.head()
```

**For LoanAmount column:**

1. Fill null values with median

```python
df_clean['LoanAmount'] = df_clean['LoanAmount'].fillna(df_clean['LoanAmount'].median())
df_clean.head()
```

**For Credit_History column:**

1. Fill null values with mode
2. Change column name to Credit

```python
df_clean['Credit_History'] = df_clean['Credit_History'].fillna(df_clean['Credit_History'].mode()[0]).astype(int)
df_clean.rename (columns={'Credit_History' : 'Credit'}, inplace = True)
df_clean.head()
```

```python
from scipy.stats import chi2_contingency

categorical_features = [
    'Female',
    'Married',
    'Dependents',
    'Graduate',
    'Self_Employed',
    'Credit',
    'Rural',
    'Semiurban',
    'Urban'
]

results = []

for col in categorical_features:

    # Create a contingency table
    table = pd.crosstab(df_clean[col], df_clean['Approved'])

    # Perform chi-square test of independence
    chi2, p_value, dof, expected = chi2_contingency(table)

    results.append({
        'Feature': col,
        'Chi2_Statistic': chi2,
        'P_Value': p_value,
        'Degrees_of_Freedom': dof,
        'Min_Expected_Frequency': expected.min()
    })

chi2_df = pd.DataFrame(results)

chi2_df = chi2_df.sort_values(
    by='P_Value'
).reset_index(drop=True)

chi2_df
```

```python
from scipy.stats import chi2_contingency

alpha = 0.05

categorical_features = [
    'Female', 'Married', 'Dependents', 'Graduate',
    'Self_Employed', 'Credit', 'Rural', 'Semiurban', 'Urban'
]

results = []

for col in categorical_features:
    table = pd.crosstab(df_clean[col], df_clean['Approved'])

    chi2, p_value, dof, expected = chi2_contingency(table)

    decision = (
        'Reject H0' if p_value < alpha
        else 'Fail to reject H0'
    )

    results.append({
        'Feature': col,
        'Chi2_Statistic': chi2,
        'P_Value': p_value,
        'Degrees_of_Freedom': dof,
        'Alpha': alpha,
        'Decision': decision
    })

chi2_df = pd.DataFrame(results).sort_values(
    'P_Value'
).reset_index(drop=True)

chi2_df
```
